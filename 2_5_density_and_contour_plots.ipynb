# **2.5 density and contour plots**

# **1. Density Plots**

Theory: Density plots show the distribution of data points across a two-dimensional space. They provide a way to visualize the concentration of data points, often using color gradients to represent areas of higher or lower density. This can be particularly useful for understanding the distribution of data in 2D.
Syntax: In Matplotlib, you can create density plots using the hist2d function or the kdeplot function from Seaborn (which uses kernel density estimation).


```python
import matplotlib.pyplot as plt
import numpy as np

# Data
x = np.random.randn(1000)
y = np.random.randn(1000)

# Density plot using hist2d
plt.hist2d(x, y, bins=30, cmap='Blues')
plt.colorbar(label='Density')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('2D Density Plot')
plt.show()
```

# **Seaborn Example:**

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Data
x = np.random.randn(1000)
y = np.random.randn(1000)

# Density plot using seaborn
sns.kdeplot(x=x, y=y, cmap='Blues', fill=True)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Kernel Density Estimate Plot')
plt.show()
```

# 2. Contour Plots

Theory: Contour plots are used to represent three-dimensional data in two dimensions. They show levels of constant value for a function over a 2D space, with lines or filled areas indicating different values of the function. This is useful for visualizing the shape and behavior of a function.
Syntax: In Matplotlib, you can create contour plots using the contour and contourf functions. contour creates contour lines, while contourf creates filled contours.


```python
import numpy as np
import matplotlib.pyplot as plt

# Define the function
def f(x, y):
    return np.sin(np.sqrt(x**2 + y**2))

# Create grid and multivariate normal
x = np.linspace(-5.0, 5.0, 100)
y = np.linspace(-5.0, 5.0, 100)
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

# Contour plot
plt.contour(X, Y, Z, levels=10, cmap='viridis')
plt.colorbar(label='Function Value')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Contour Plot')
plt.show()
```

# **Filled Contour Plot Example:**

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the function
def f(x, y):
    return np.sin(np.sqrt(x**2 + y**2))

# Create grid and multivariate normal
x = np.linspace(-5.0, 5.0, 100)
y = np.linspace(-5.0, 5.0, 100)
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

# Filled contour plot
plt.contourf(X, Y, Z, levels=20, cmap='viridis')
plt.colorbar(label='Function Value')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Filled Contour Plot')
plt.show()
```

# **Visualizing a Three-Dimensional Function**

Theory: Visualizing a three-dimensional function involves plotting a surface or mesh that represents the function's values over a 2D grid. This can be useful for understanding the shape of the function and its behavior in 3D space.
Syntax: In Matplotlib, you can use the plot_surface function from the mpl_toolkits.mplot3d module to create 3D surface plots


```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define the function
def f(x, y):
    return np.sin(np.sqrt(x**2 + y**2))

# Create grid
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

# Create 3D plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')

ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')
ax.set_title('3D Surface Plot')
plt.show()
```

# *2.6 Histograms*

•	Histograms: A histogram is a graphical representation of data distribution. It divides the data range into intervals, called bins, and counts the number of data points that fall into each bin. The height of each bar represents the frequency of data points within that bin.

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
data = np.random.randn(1000)

# Plot histogram
plt.hist(data, bins=30, density=False, alpha=0.75, color='blue')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram with 30 Bins')
plt.show()
```

# **Two-Dimensional Histograms and Binning**

Theory:
•	Two-Dimensional Histograms: A 2D histogram is an extension of a 1D histogram that visualizes the joint distribution of two variables. Instead of bars, the data is divided into 2D bins (rectangles), with the height of each bin representing the count of data points within that bin.
•	2D Binning: Similar to 1D binning, but applied to both dimensions. It involves dividing the x and y ranges into bins, creating a grid of bins. The color intensity of each bin typically represents the count of data points in that bin.
Syntax: In Matplotlib, you can create 2D histograms using the hist2d function:
plt.hist2d(x, y, bins=30, cmap='Blues')
plt.colorbar()  # To show the color scale
•	x and y: Input data for the two dimensions.
•	bins: Number of bins for both x and y dimensions.
•	cmap: Colormap for the bins.


```python
import matplotlib.pyplot as plt
import numpy as np

# Data
x = np.random.randn(1000)
y = np.random.randn(1000)

# Plot 2D histogram
plt.hist2d(x, y, bins=30, cmap='Blues')
plt.colorbar(label='Counts')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('2D Histogram with 30x30 Bins')
plt.show()
```

# **2. Legend for Size of Points**

Theory: When plotting data points of varying sizes, it's often useful to include a legend that explains what different point sizes represent. This is particularly important in scatter plots where size may encode additional information, like population, magnitude, or importance.
Syntax: You can manually create a legend for point sizes using plt.scatter and plt.legend. The handles and labels parameters of plt.legend allow you to customize the legend entries.

plt.scatter(x, y, s=sizes, label='Data')
handles, labels = plt.gca().get_legend_handles_labels()
plt.legend(handles, labels)


```python
import matplotlib.pyplot as plt
import numpy as np

# Data
x = np.linspace(0, 10, 50)
y = np.sin(x)
sizes = np.linspace(50, 500, 50)  # Variable point sizes

# Scatter plot with size legend
scatter = plt.scatter(x, y, s=sizes, alpha=0.5, label='Data Points')
legend_sizes = [50, 200, 500]
for size in legend_sizes:
    plt.scatter([], [], s=size, alpha=0.5, label=f'Size {size}')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Legend for Size of Points')
plt.legend(scatterpoints=1, frameon=True, labelspacing=1, title="Point Sizes")
plt.show()
```

# **3. Multiple Legends**

Theory: In complex plots with multiple datasets or visual elements, you might need more than one legend to maintain clarity. Matplotlib allows the creation of multiple legends by using the Legend class directly.
Syntax: To add multiple legends, you create additional legend instances using plt.gca().add_artist(legend).

legend1 = plt.legend(handles, labels, loc='upper left')
plt.gca().add_artist(legend1)


```python
import matplotlib.pyplot as plt
import numpy as np

# Data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# Plot with multiple legends
line1, = plt.plot(x, y1, color='blue', label='Sine Wave')
line2, = plt.plot(x, y2, color='green', label='Cosine Wave')

# First legend
legend1 = plt.legend(handles=[line1], loc='upper left', title='Wave 1')
plt.gca().add_artist(legend1)

# Second legend
legend2 = plt.legend(handles=[line2], loc='lower right', title='Wave 2')

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Multiple Legends')
plt.show()
```

*2.8 colors *

Colorbars are essential when visualizing data that includes a range of values mapped to colors, such as in heatmaps, contour plots, or image data. A colorbar provides a reference that indicates how the data values correspond to the colors used in the plot. Customizing colorbars allows you to control the appearance, orientation, tick marks, labels, and colormap used in the plot.
When visualizing something like handwritten digits, colorbars can help interpret the pixel intensity values by mapping them to a color gradient.
Key Elements of Customization:
1.	Colormap (cmap): Specifies the range of colors used. Matplotlib offers many predefined colormaps.
2.	Orientation: The colorbar can be placed vertically or horizontally.
3.	Ticks and Labels: Control the placement and labeling of ticks on the colorbar.
4.	Size and Location: Adjust the size and position of the colorbar relative to the plot.
5.	Extend: Allows extending the colorbar beyond the highest or lowest data values.
Syntax: To add and customize a colorbar in Matplotlib, the syntax generally involves:


```python
import matplotlib.pyplot as plt
import numpy as np

# Create sample 2D data (like a heatmap)
data = np.random.randint(0, 256, (10, 10))   # values from 0 to 255

# Display the heatmap
plt.imshow(data, cmap='viridis')

# Add a colorbar
cbar = plt.colorbar()

# Customize the colorbar
cbar.set_label('Intensity')
cbar.set_ticks([0, 50, 100, 150, 200, 255])
cbar.set_ticklabels(['Min', 'Low', 'Medium', 'High', 'Very High', 'Max'])

# Add title
plt.title("Heatmap with Customized Colorbar")

plt.show()
```

# Example: Visualizing Handwritten Digits with a Customized Colorbar

Handwritten digits can be visualized using the MNIST dataset, where each digit is a 28x28 pixel grayscale image. We’ll display a digit and customize the colorbar to better interpret the intensity values.

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.datasets import load_digits

# Load a sample digit from the dataset
digits = load_digits()
digit_image = digits.images[0]  # Load the first digit (a '0')

# Plot the image
plt.imshow(digit_image, cmap='gray')

# Add a customized colorbar
cbar = plt.colorbar()
cbar.set_label('Pixel Intensity')
cbar.set_ticks([0, 50, 100, 150, 200, 255])
cbar.set_ticklabels(['0', 'Low', 'Medium', 'High', 'Very High', '255'])
cbar.ax.invert_yaxis()  # Invert the y-axis if needed

plt.title('Handwritten Digit with Customized Colorbar')
plt.show()
```

**2.9 subplots**
When creating visualizations, it’s often useful to display multiple plots in a single figure. Matplotlib offers several ways to create subplots, ranging from simple to more complex layouts.


# **1. plt.axes: Subplots by Hand**

```python
import matplotlib.pyplot as plt

# Create a figure
plt.figure()

# First subplot
ax1 = plt.axes([0.1, 0.1, 0.4, 0.4])
ax1.plot([0, 1], [0, 1])
ax1.set_title('First Subplot')

# Second subplot
ax2 = plt.axes([0.6, 0.1, 0.3, 0.3])
ax2.plot([0, 1], [1, 0])
ax2.set_title('Second Subplot')

plt.show()
```

# **2. plt.subplot: Simple Grids of Subplots**

Theory: plt.subplot is used to create a grid of subplots within a figure. This method is simpler than plt.axes and is suitable for evenly spaced grids.
Syntax:
plt.subplot(nrows, ncols, index)
•	nrows, ncols: Number of rows and columns in the grid.
•	index: The position of the subplot (1-based index).


```python
import matplotlib.pyplot as plt

# Create a figure
plt.figure()

# First subplot in a 2x2 grid
plt.subplot(2, 2, 1)
plt.plot([0, 1], [0, 1])
plt.title('Subplot 1')

# Second subplot in a 2x2 grid
plt.subplot(2, 2, 2)
plt.plot([0, 1], [1, 0])
plt.title('Subplot 2')

# Third subplot in a 2x2 grid
plt.subplot(2, 2, 3)
plt.plot([0, 1], [0.5, 0.5])
plt.title('Subplot 3')

# Fourth subplot in a 2x2 grid
plt.subplot(2, 2, 4)
plt.plot([0, 1], [0.5, 1])
plt.title('Subplot 4')

plt.show()
```

# **3. plt.subplots: The Whole Grid in One Go**

Theory: plt.subplots is a more powerful and flexible method that creates an entire grid of subplots in one command. It returns both the figure and an array of axes, which makes it easy to iterate over subplots.
Syntax:
fig, ax = plt.subplots(nrows, ncols, figsize=(width, height))
•	nrows, ncols: Number of rows and columns.
•	figsize: Tuple specifying the width and height of the figure


```python
import matplotlib.pyplot as plt

# Create a figure and a grid of subplots
fig, ax = plt.subplots(2, 2, figsize=(8, 6))

# First subplot
ax[0, 0].plot([0, 1], [0, 1])
ax[0, 0].set_title('Subplot 1')

# Second subplot
ax[0, 1].plot([0, 1], [1, 0])
ax[0, 1].set_title('Subplot 2')

# Third subplot
ax[1, 0].plot([0, 1], [0.5, 0.5])
ax[1, 0].set_title('Subplot 3')

# Fourth subplot
ax[1, 1].plot([0, 1], [0.5, 1])
ax[1, 1].set_title('Subplot 4')

plt.tight_layout()
plt.show()
```

# **4. plt.GridSpec: More Complicated Arrangements**

```python
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

# Create a figure
plt.figure()

# Create a GridSpec
gs = gridspec.GridSpec(3, 3)

# First subplot spans all columns of the first row
ax1 = plt.subplot(gs[0, :])
ax1.plot([0, 1], [0, 1])
ax1.set_title('Subplot 1: Full Row')

# Second subplot spans the first two rows and the last column
ax2 = plt.subplot(gs[1:, -1])
ax2.plot([0, 1], [1, 0])
ax2.set_title('Subplot 2: Bottom Right')

# Third subplot spans the first and second columns of the second and third rows
ax3 = plt.subplot(gs[1:, :-1])
ax3.plot([0, 1], [0.5, 0.5])
ax3.set_title('Subplot 3: Bottom Left')

plt.tight_layout()
plt.show()
```


# **1. Effect of Holidays on US Births**
Theory: Analyzing the effect of holidays on US births involves overlaying textual annotations or markers on a plot to indicate specific dates or holidays. This can help to visualize patterns, such as whether certain holidays influence birth rates.
Example: Let's assume you have a dataset showing daily births in the US, and you want to highlight the effects of major holidays like Christmas and New Year.


```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Simulated data: Daily births
dates = pd.date_range(start='2022-12-20', end='2023-01-10')
births = np.random.poisson(100, len(dates)) + 20 * (dates.day == 25)  # Spike on Christmas

# Plotting the data
plt.plot(dates, births, label='Daily Births')

# Get exact index positions for dates
christmas_date = pd.Timestamp('2022-12-25')
newyear_date = pd.Timestamp('2023-01-01')

christmas_index = dates.get_loc(christmas_date)
newyear_index = dates.get_loc(newyear_date)

# Highlighting Christmas
plt.annotate(
    'Christmas',
    xy=(christmas_date, births[christmas_index]),
    xytext=(pd.Timestamp('2022-12-22'), births[christmas_index] + 30),
    arrowprops=dict(facecolor='black', shrink=0.05)
)

# Highlighting New Year
plt.annotate(
    'New Year',
    xy=(newyear_date, births[newyear_index]),
    xytext=(pd.Timestamp('2022-12-29'), births[newyear_index] + 30),
    arrowprops=dict(facecolor='black', shrink=0.05)
)

plt.xlabel('Date')
plt.ylabel('Number of Births')
plt.title('Effect of Holidays on US Births')
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
